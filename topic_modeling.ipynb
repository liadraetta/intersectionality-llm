```python
from bertopic import BERTopic
```

```python
from collections import defaultdict
from glob import glob 
import os 
import pandas as pd 

filter_type = "overall"

folder =f"embeddings/{filter_type}"
pattern = "clusters_*.csv"

for file in glob(os.path.join(folder, pattern)):
    print(file)
    df = pd.read_csv(file)
    filename=file.split("/")[-1]
    input_name = filename.split(".")[0]

    cluster_topics = defaultdict(list)  # to store topics per cluster

    for cluster_id in df['cluster'].unique():
        cluster_texts = df[df['cluster'] == cluster_id]['reasoning'].tolist()
        
        if len(cluster_texts) < 2:
            print(f"Cluster {cluster_id} has too few texts for topic modeling. Skipping.")
            continue

        # Fit BERTopic on this cluster
        topic_model = BERTopic()
        topics, probs = topic_model.fit_transform(cluster_texts)
        
        print(f"\n=== Cluster {cluster_id} Topics ===")
        cluster_topics[cluster_id] = topic_model.get_topic_info()
        print(cluster_topics[cluster_id].head(5))  # top 5 topics

        for cluster_id, topic_info in cluster_topics.items():
            topic_info.to_csv(f"embeddings/{filter_type}/clusters/{input_name}_{cluster_id}_topics.csv", index=False)
```
