##### Clean and add demographics wherever missing.

```python
import pathlib
import pandas as pd
def get_demographics(original_data, annId, demographic_traits=None):
    user_demographics = original_data[original_data['annId'] == annId]
    
    demographic_traits_to_column = {
        'gender': 'annotatorGender',
        'race': 'annotatorRace',
        'political': 'annotatorPolitics'
    }

    if demographic_traits is None:
        return {}
    elif isinstance(demographic_traits, str):
        output_traits =  {demographic_traits: user_demographics[demographic_traits_to_column[demographic_traits]].values[0]}
    elif isinstance(demographic_traits, list):
        output_traits = {trait: user_demographics[demographic_traits_to_column[trait]].values[0] for trait in demographic_traits}
    else:
        raise ValueError("demographic_traits must be a str or a list")
    
    # Postprocess political to be discrete
    if 'political' in output_traits:
        if output_traits['political'] > 0:
            output_traits['political'] = 'conservative'
        elif output_traits['political'] < 0:
            output_traits['political'] = 'liberal'
        else:
            output_traits['political'] = 'neutral'

    return output_traits
```

```python
model_names = ['deepseek', 'Llama', 'Qwen2', 'gemma', 'Ministral']
original_data_path = pathlib.Path('../dataset/AnnAttDemographics.csv')
original_data = pd.read_csv(original_data_path)
```

```python
import os
for model_name in model_names:
    results_cleaned_path = pathlib.Path(f"{model_name}/cleaned")
    for curr_df_path in os.listdir(results_cleaned_path):
        # Extract socio-demographic traits from filename
        if 'gender_race_political' in curr_df_path:
            socio_demographic_traits = ['gender', 'race', 'political']
        elif 'gender_race' in curr_df_path:
            socio_demographic_traits = ['gender', 'race']
        elif 'gender_political' in curr_df_path:
            socio_demographic_traits = ['gender', 'political']
        elif 'race_political' in curr_df_path:
            socio_demographic_traits = ['race', 'political']
        elif 'race' in curr_df_path:
            socio_demographic_traits = 'race'
        elif 'gender' in curr_df_path:
            socio_demographic_traits = 'gender'
        elif 'political' in curr_df_path:
            socio_demographic_traits = 'political'
        else:
            raise ValueError("NOT FOUND")
        curr_df = pd.read_csv(results_cleaned_path / curr_df_path)
        curr_df_demographics = []
        for idx, row in curr_df.iterrows():
            annId = row['annId']
            demographics = get_demographics(original_data, annId, socio_demographic_traits)
            curr_df_demographics.append(demographics)
        curr_df['demographics'] = curr_df_demographics
        curr_df.to_csv(results_cleaned_path / curr_df_path)
        print(f"Processed {curr_df_path}")
```
